```python
%sql
create or replace view  IDENTIFIER(:catalog || '.' || 'naval_gold.top_5_customers') as 
select customer_id, sum(total_amount) as total_sales from  IDENTIFIER(:catalog || '.' || 'naval_silver.sales_cleaned') group by customer_id order by total_sales desc limit 5
```

```python
%sql
create or replace view  IDENTIFIER(:catalog || '.' || 'naval_gold.top_5_products') as 
select product_id, round(sum(total_amount)) as total_sales from  IDENTIFIER(:catalog || '.' || 'naval_silver.sales_cleaned') group by product_id order by total_sales desc limit 5
```
