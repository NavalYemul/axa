```python
schema= "order_id int ,customer_id int ,transaction_id int ,product_id int ,quantity int ,discount_amount double ,total_amount double ,order_date string"
```

```python
# Streaming
df=spark.readStream.csv("/Volumes/dev/naval/raw/months/",header=True,schema=schema)
df.writeStream.option("checkpointLocation","/Volumes/dev/naval/raw/checkpoints/").trigger(availableNow=True).toTable("dev.naval_bronze.sales_stream")
```
